```python
#Assignment2 : using titanic data set find total no of passengers and find average age of survivors and non survivors and list the passengers who are older than 60 years find the average fare for each passenger class

import pandas as pd

# Load the standard Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# 1. Total number of passengers
total_passengers = df.shape[0]  # or len(df)
print(f"Total Passengers: {total_passengers}")

# 2. Average age of survivors and non-survivors
# 1 = Survived, 0 = Died
avg_age_by_survival = df.groupby("Survived")["Age"].mean()
print("\nAverage Age by Survival Status:")
print(avg_age_by_survival)

# 3. List passengers older than 60 years
older_than_60 = df[df["Age"] > 60]
print(f"\nPassengers older than 60 (Total: {len(older_than_60)}):")
print(older_than_60[["Name", "Age", "Pclass"]].head())

# 4. Average fare for each passenger class
avg_fare_by_class = df.groupby("Pclass")["Fare"].mean()
print("\nAverage Fare for Each Passenger Class:")
print(avg_fare_by_class)
```
